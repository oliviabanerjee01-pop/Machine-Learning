```python
w_values = np.linspace(0, 4, 100)
cost_values = []

for w in w_values:
    predictions = w * x + b
    cost = np.mean((predictions - y)**2) / 2
    cost_values.append(cost)

plt.plot(w_values, cost_values)
plt.xlabel("w")
plt.ylabel("Cost J(w)")
plt.title("Cost function")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Training data
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 8, 10])

# Possible values of w and b
w_values = np.linspace(0, 4, 50)
b_values = np.linspace(-5, 5, 50)

# Create a grid of all w and b combinations
W, B = np.meshgrid(w_values, b_values)

# Empty array to store the cost for every combination
J = np.zeros_like(W)

# Calculate cost for every w and b combination
for i in range(len(b_values)):
    for j in range(len(w_values)):

        # Make predictions
        predictions = W[i, j] * x + B[i, j]

        # Calculate cost
        J[i, j] = np.mean((predictions - y) ** 2) / 2

# Create 3D graph
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

# Plot cost surface
ax.plot_surface(W, B, J)

# Labels
ax.set_xlabel("w")
ax.set_ylabel("b")
ax.set_zlabel("Cost J(w,b)")
ax.set_title("Cost Function J(w,b)")

plt.show()
```
